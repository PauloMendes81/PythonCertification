```python
inventory = {"laptops": 10, "mice": 25}


def add_item(current_inv, item, quantity):
    if item in current_inv:
        current_inv[item] += quantity
    else:
        current_inv[item] = quantity


def view_inventory(current_inv):
    print("--- Current Stock ---")
    for product, count in current_inv.items():
        print(f"{product}: {count}")


def total_items(current_inv):
    all_numbers = current_inv.values()
    sum_value = sum(all_numbers)
    print(f"Total items in store: {sum_value}")


add_item(inventory, "Laptops", 5)
add_item(inventory, "Keyboards", 12)
view_inventory(inventory)
total_items(inventory)
```
