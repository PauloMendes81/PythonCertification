```python
class Planet:
    def __init__(self, name, planet_type, star):
        # 1. Type Validation
        if not all(isinstance(attr, str) for attr in [name, planet_type, star]):
            raise TypeError("name, planet type, and star must be strings")

        # 2. Empty String Validation
        if not all(len(attr) > 0 for attr in [name, planet_type, star]):
            raise ValueError("name, planet_type, and star must be non-empty strings")

        # 3. Assignment
        self.name = name
        self.planet_type = planet_type
        self.star = star

    def orbit(self):
        return f"{self.name} is orbiting around {self.star}..."

    def __str__(self):
        return f"Planet: {self.name} | Type: {self.planet_type} | Star: {self.star}"

# Creating three instances with data
planet_1 = Planet("Earth", "Terrestrial", "The Sun")
planet_2 = Planet("Jupiter", "Gas Giant", "The Sun")
planet_3 = Planet("Proxima Centauri b", "Super-Earth", "Proxima Centauri")

# Printing the objects (__str__ is called automatically by print)
print(planet_1)
print(planet_2)
print(planet_3)

# Calling and printing the orbit method
print(planet_1.orbit())
print(planet_2.orbit())
print(planet_3.orbit())
```
