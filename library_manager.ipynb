```python
library = {
    "The Great Gatsby": ["available", ["Jack", "Rosie", "David"]],
    "On the Origin of Species": ["borrowed", ["Joseph"]],
    "Lord Of The Rings": ["available", ["Sergio", "Matt"]]
}

def borrow_book(catalog, title, person_name):
    book = catalog[title] # This is the list [status, history]

    # Check index 0 for status
    if book[0] == "available":
        book[0] = "borrowed"
        book[1].append(person_name) # Append to the list at index 1
        print(f"Have fun with your book, {person_name}!")
    else:
        print(f"Sorry {person_name}, '{title}' is already borrowed.")

def return_book(catalog, title):
    book = catalog[title]
    if book[0] == "borrowed":
        book[0] = "available"
        print(f"'{title}' has been returned and is now available.")

def search_history(catalog, title):
    # Grab the history list at index 1
    history = catalog[title][1]
    print(f"History for '{title}': {', '.join(history)}")


borrow_book(library, "On the Origin of Species", "Jack") # Should fail (already borrowed)
return_book(library, "On the Origin of Species")         # Now it's free
borrow_book(library, "On the Origin of Species", "Jack") # Now it works!
search_history(library, "The Great Gatsby")
```

The Challenge: The Library Manager

You are in charge of a library. You need to keep track of which books are available and who has borrowed them.

1. The Data Structure
Create a dictionary called library.

Keys: Book titles (e.g., "The Great Gatsby")

Values: A Dictionary containing:

"status": Either "available" or "borrowed".

"history": A List of names of people who have borrowed it.

2. The Functions
borrow_book(catalog, title, person_name):

Find the book in the catalog.

Change its status to "borrowed".

Append the person's name to the book's "history" list.

return_book(catalog, title):

Change the status back to "available".

search_history(catalog, title):

Print the list of everyone who has ever read that book.
