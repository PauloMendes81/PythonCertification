```python
grades = {"Alice": 85, "Bob": 72}

def add_student(data, name, score):
    if name in data:
        print(f'the student {name} has alredy been added.')
    else:
        data[name] = score

def check_pass_fail(data, name):
    student_score = data[name]

    if student_score >= 60:
        print(f'the student {name} Passed!')
    else:
        print(f'the student {name} Failed...')

def calculate_average(data):
    student_values = data.values()
    average_score = sum(student_values) / len(student_values)
    print(f"The average score for this class is {average_score:.2f}")

add_student(grades, "Charlie", 55)
check_pass_fail(grades, "Alice")
check_pass_fail(grades, "Charlie")
calculate_average(grades)
```
