```python
tracker = {"Alex": [30, 45, 20], "Jamie": [60]}

def add_workout(data, name, minutes):
    if name in data:
        data[name].append(minutes)
    else:
        data[name]= [minutes]

def show_stats(data, name):
    workouts = data[name]

    total_minutes = sum(workouts)
    number_sessions = len(workouts)

    print(f"{name}'s total minutes: {total_minutes}")
    print(f"{name} has worked out {number_sessions} times.")

def heavy_hitters(data):
    print("--- Heavy Hitters (100+ mins) ---")

    for name, minutes_list in data.items():
        if sum(minutes_list) >= 100:
            print(f"{name} is a heavy hitter!")

add_workout(tracker, "Jamie", 50)
show_stats(tracker, "Jamie")
heavy_hitters(tracker)
```

The Challenge: Fitness Tracker

Imagine you are building an app to track how many minutes someone exercises. Instead of just one number, we want to keep a list of all their workout times.

Your Goal:
The Data: Create a dictionary called tracker where the keys are names and the values are lists of numbers.

Example: {"Alex": [30, 45, 20], "Jamie": [60]}

Function 1: add_workout(data, name, minutes)

If the name exists, .append() the minutes to their list.

If the name doesn't exist, create a new key and set its value to a list containing that one number: [minutes].

Function 2: show_stats(data, name)

Calculate the total minutes (sum) and the number of workouts (len) for that person.

Function 3: heavy_hitters(data)

Find any person who has worked out for a total of more than 100 minutes.
