```python
match_results = ["Alex", "Jamie", "Alex", "Sam", "Alex", "Jamie"]

def count_goals(goal_list):
    score_board = {}
    for player in goal_list:
        if player in score_board:
            score_board[player] += 1
        else:
            score_board[player] = 1
    return score_board

def show_winner(score_board):
    max_score = None
    winner_name = ""

    for player in score_board:
        current_value = score_board[player]

        if max_score is None or current_value > max_score:
            max_score = current_value
            winner_name = player # Save the name of the person with high score

    print(f"The greatest score is: {max_score} goals by {winner_name}!")

final_scores = count_goals(match_results)
show_winner(final_scores)
```
