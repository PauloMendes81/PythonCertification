```python
prices = {"Laptop": 1000, "Mouse": 50, "Monitor": 300}

def apply_discount(data, percent):
    for product in data:
        price = data[product]

        # Calculate the amount to take off
        discount_amount = (price * percent) / 100

        # Update the dictionary at that specific key
        data[product] = price - discount_amount

# Applying a 10% discount
apply_discount(prices, 10)

print(prices)
```

The Challenge: The "Price Is Right"

Imagine you have a dictionary of products and their prices. You want to write a function that applies a discount to all of them.

The Data: prices = {"Laptop": 1000, "Mouse": 50, "Monitor": 300}

The Function: apply_discount(data, percent)

This function should loop through the dictionary.

It should reduce every price by the percent (e.g., if percent is 10, the Laptop becomes 900).

The Catch: This time, don't return anything. Just modify the dictionary and then print it outside.
